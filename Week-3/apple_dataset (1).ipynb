```python
df=df.sort_values("Date")
```

```python
avg=df[["Close","Open"]].mean()
max=df[["Close","Open"]].max()
min=df[["Close","Open"]].min()
print(avg,max,min)
```

```python
threshold=2*avg
print(threshold)
```

```python
df["Daily_delta"] = df["Close"]-df["Open"]
top_10_positive = (df[df["Daily_delta"] > 0].nlargest(10, "Daily_delta"))
print(top_10_positive)
```

```python
df["daily_return"] = df["Close"]-df["Open"]
df["daily_return"] = df["daily_return"]/df["Open"] *100
print(df["daily_return"])
```

```python
highest_10_returns = (df[df["daily_return"].notna()].nlargest(10, "daily_return"))
print(highest_10_returns)
```

```python
lowest_10_returns = (df[df["daily_return"].notna()].nsmallest(10, "daily_return"))
print(lowest_10_returns)
```

```python
anolomous_days=df[df["Close"] > threshold["Close"]]
print(anolomous_days)
```

```python
print(df[["Date", "Open", "Close", "Daily_delta", "daily_return"]].head())
```

```python
print("\nAnomalous Trading Days:")
print(anolomous_days[["Date","Open","Close","Daily_delta","daily_return","Volume"]])
```
